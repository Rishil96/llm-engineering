```python
# Import libraries
import os
import gradio as gr
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# Initialization
load_dotenv(override=True)

openai_api_key = os.getenv("OPENAI_API_KEY")
if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")

MODEL = "gpt-4o-mini"
openai = OpenAI()
```

```python
# System prompt
system_prompt = """
You are an expert technical tutor capable of answering any questions related to programming.
You will be asked programming concepts related questions or asked to debug code that has errors.
Your task is to provide an explanation which is easy to understand using examples.
For code debugging queries, simply provide the solution and explain what was wrong with the code.
Always answer in markdown.
"""
```

```python
# Chat function to be used by Gradio
def chat(message, history, model):
    """
    message: latest user message
    history: history of the conversation
    model: model to use for chat
    """
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(
        model=model,
        messages=messages,
    )
    return response.choices[0].message.content
    
```

```python
# Create Gradio Interface
models = ["gpt-3.5-turbo", "gpt-4o-mini"]

with gr.Blocks() as ui:
    model = gr.Dropdown(label="Select Model", choices=models, value=models[0])
    chatbot = gr.Chatbot(type="messages")
    msg = gr.Textbox(placeholder="Type your message...")
    send_btn = gr.Button("Send")

    def handle_click(message, chat_history, selected_model):
        return chat(message, chat_history, selected_model)

    send_btn.click(
        fn=handle_click,
        inputs=[msg, chatbot, model],
        outputs=chatbot
    )

ui.launch()
```
