```python
import os
from huggingface_hub import login
from transformers import AutoTokenizer
from dotenv import load_dotenv
```

```python
# Login to Huggingface using Access Key
load_dotenv()
hf_token = os.getenv("HF_TOKEN")
login(hf_token, add_to_git_credential=True)
```

```python
# Create tokenizer for the model we want to use
tokenizer = AutoTokenizer.from_pretrained(
    "meta-llama/Meta-Llama-3.1-8B", 
    trust_remote_code=True
)
```

```python
# Encode text using llama tokenizer
text = "I am excited to show Tokenizers in action to my LLM engineers"
tokens = tokenizer.encode(text)
tokens
```

```python
# Decode tokens to text
tokenizer.decode(tokens)
tokenizer.batch_decode(tokens)
```

```python
# To check vocab of tokenizer
tokenizer.vocab
# To check special tokens used by LLM
tokenizer.get_added_vocab()
```

```python
# Instruct variant of models
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Meta-Llama-3.1-8B-Instruct", trust_remote_code=True)
```

```python
# Create chat history
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Tell me a light-hearted joke for a room of Data Scientists."}
]

# Use method to restructure the messages using special tokens for LLM to understand.
prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
print(prompt)
```
